# Limpieza de datos

## Acción en variables categóricas
### Razón de las columnas eliminadas:
### No aportan valor predictivo:
**id**: Es únicamente un identificador numérico de fila.

**recorded_by**: Solo tiene 1 único valor para todos los pozos del dataset ("GeoData Consultants Ltd").

**scheme_name**: Contiene casi un 50% de valores nulos y altísima cardinalidad (2.695 nombres diferentes). Aporta demasiado ruido y rellenar esa cantidad de nulos sería inventarse la mitad de la información.

**subvillage**: Decidimos borrar la columna por completo debido a su extrema cardinalidad (19.287 pueblos distintos). Aporta demasiado ruido geográfico y ya tenemos la ubicación cubierta con la latitud/longitud y la región.

### Limpieza por repetidas o mejor explicadas:

**extraction_type** (18 valores) y **extraction_type_group** (13 valores): Aportan la misma información sobre el método de extracción pero con demasiado detalle. Se eliminan en favor de extraction_type_class, que resume lo mismo en solo 7 categorías perfectas para el modelo.

**region_code** y **district_code**: Son códigos numéricos redundantes de la ubicación geográfica. Al ser números, el modelo puede interpretarlos erróneamente como valores continuos (creyendo que la región 20 "vale más" que la 10). La ubicación ya está bien cubierta con la columna de texto region y las coordenadas.

**management** Esta más resumido en la columna management_group 

**payment_type**: Es un duplicado exacto de la columna payment (ambas tienen los mismos 7 valores).

**water_quality**: Tiene demasiadas categorías pequeñas con muy pocos datos (ej. "fluoride abandoned" con solo 17 filas), lo que provoca sobreajuste. Se elimina en favor de quality_group, que agrupa estos detalles en bloques más representativos.

**quantity_group**: Es otro duplicado matemáticamente exacto de la columna quantity (ambas distribuyen la información en las mismas 5 categorías).

**source** (10 valores) y **source_class** (3 valores): Son redundantes. source tiene demasiado nivel de detalle (categorías con menos de 50 registros) y source_class es demasiado genérica. Se eliminan para quedarse con la opción intermedia y más equilibrada: source_type.

**waterpoint_type**: Es prácticamente idéntica a waterpoint_type_group, pero esta última agrupa de forma más limpia las categorías minoritarias, reduciendo el número de columnas a la hora de hacer el One-Hot Encoding.

## Columnas con nulos:
**subvillage (371 nulos)**: Eliminar. Aunque tiene muy pocos nulos, decidimos borrar la columna por completo debido a su extrema cardinalidad (19.287 pueblos distintos). Aporta demasiado ruido geográfico y ya tenemos la ubicación cubierta con la latitud/longitud y la región.

**funder (3.637 nulos)** e **installer (3.655 nulos)**: Imputar con 'Desconocido'. Al ser variables de texto con muchísimos valores, los algoritmos fallarían si dejamos los vacíos.

**scheme_management (3.878 nulos)**: Imputar con 'Other'. Es una variable de baja cardinalidad (solo 11 valores) que convenientemente ya incluye la categoría "Other" (Otros). Aprovechamos esta categoría existente para absorber los valores faltantes de una forma muy limpia para el negocio.

**public_meeting (3.334 nulos)** y **permit (3.056 nulos)**: Imputar con la Moda (True). Ambas son variables booleanas (Verdadero o Falso). Al observar los datos, la inmensa mayoría de los pozos sí tienen reuniones públicas y sí tienen permiso. Por estadística básica, aplicamos la moda y asumimos que los que no tienen el dato registrado probablemente también sean True.

```python
def limpiar_datos_cat(df):

    df_clean = df.copy()
    
    # 1. Eliminar columnas inútiles
    cols_drop = ['id', 'subvillage', 'recorded_by', 'scheme_name', 'extraction_type', 'extraction_type_group', 'region_code', 'district_code',
                 'management', 'payment_type', 'water_quality', 'quantity_group', 'source_class', 'source', 'waterpoint_type' ]
    df_clean = df_clean.drop(columns=cols_drop)
    
    # 2. Imputar nulos
    # Textos de alta cardinalidad: Rellenar con 'unknown'
    df_clean['funder'] = df_clean['funder'].fillna('Unknown')
    df_clean['installer'] = df_clean['installer'].fillna('Unknown')
    df_clean['scheme_management'] = df_clean['scheme_management'].fillna('Other')
    df_clean['public_meeting'] = df_clean['public_meeting'].fillna(True)
    df_clean['permit'] = df_clean['permit'].fillna(True)
    
    # Texto de baja cardinalidad: Lo metemos en la categoría 'Other' que ya existe
    df_clean['scheme_management'] = df_clean['scheme_management'].fillna('Other')
    
    # Booleanos: Rellenar con el valor más probable (La moda: True)
    df_clean['public_meeting'] = df_clean['public_meeting'].fillna(True)
    df_clean['permit'] = df_clean['permit'].fillna(True)


    
    return df_clean
```

# Acciones Variables Numéricas
## Valores intactos:
**amount_tsh**: 
**gps_height** Min: -90. Puede ser posible que esté a la altura por debajo del mar, por lo tanto se deja.

## Valores anómalos:
**longitude y latitude**:  Se vuelve aplicar los filtros anteriores para cuando se tenga que meter información nueva.

**construction_year** y **population**: Min: 0, 25%: 0. Tienen exactamente el mismo patrón de datos faltantes sistemáticos (regiones enteras como Mbeya o Kagera registran exclusivamente ceros). Para evitar sesgos por muestras pequeñas, se aplica una **imputación híbrida**: se usa la mediana de la región solo si esta cuenta con un mínimo histórico fiable (> 150 registros válidos); en caso contrario, se imputa la mediana global.

## Columnas añadidas:
**has_amount_tsh**: Menciona si tiene presión medida o no.


## Columnas eliminadas:
**num_private**: No aporta valor.

```python
def limpiar_datos_num(df):

    df_clean = df.copy()
    umbral_confianza = 150
    
    #Creación de tablas
    df_clean['has_amount_tsh'] = (df_clean['amount_tsh'] > 0).astype(int)
    df_clean['has_construction_year'] = (df_clean['construction_year'] > 0).astype(int)
    df_clean['has_population'] = (df_clean['population'] > 0).astype(int)
    
    #Pasar los 0 a nulos y de nulos a lo mencionado anteriomente
    #construction_year
    df_clean['construction_year'] = df_clean['construction_year'].replace(0, np.nan)
    conteos_year = df_clean.groupby('region')['construction_year'].count()
    regiones_fiables_year = conteos_year[conteos_year >= umbral_confianza].index
    
    falta_year = df_clean['construction_year'].isna()
    es_fiable_year = df_clean['region'].isin(regiones_fiables_year)
    
    df_clean.loc[falta_year & es_fiable_year, 'construction_year'] = df_clean.groupby('region')['construction_year'].transform('median')
    df_clean.loc[falta_year & ~es_fiable_year, 'construction_year'] = df_clean['construction_year'].median()
    #Population

    df_clean['population'] = df_clean['population'].replace(0, np.nan)
    conteos_pop = df_clean.groupby('region')['population'].count()
    regiones_fiables_pop = conteos_pop[conteos_pop >= umbral_confianza].index
    
    falta_pop = df_clean['population'].isna()
    es_fiable_pop = df_clean['region'].isin(regiones_fiables_pop)
    
    df_clean.loc[falta_pop & es_fiable_pop, 'population'] = df_clean.groupby('region')['population'].transform('median')
    df_clean.loc[falta_pop & ~es_fiable_pop, 'population'] = df_clean['population'].median()
    

    # Latitud y Longitud: Reemplazar, aplicar media
    
    df_clean['longitude'] = pd.to_numeric(df_clean['longitude'], errors='coerce')
    df_clean['latitude'] = pd.to_numeric(df_clean['latitude'], errors='coerce')

    df_clean['longitude'] = df_clean['longitude'].replace(0.0, np.nan)
    df_clean['latitude'] = df_clean['latitude'].replace(0.0, np.nan)

    df_clean['longitude'] = df_clean['longitude'].fillna(df_clean.groupby('region')['longitude'].transform('median'))
    df_clean['latitude'] = df_clean['latitude'].fillna(df_clean.groupby('region')['latitude'].transform('median'))

    df_clean['longitude'] = df_clean['longitude'].fillna(df_clean['longitude'].median())
    df_clean['latitude'] = df_clean['latitude'].fillna(df_clean['latitude'].median())
    
    #Añadimos columna:
    df_clean['has_amount_tsh'] = (df_clean['amount_tsh'] > 0).astype(int)

    #Eliminamos columna:
    df_clean = df_clean.drop(columns=['num_private'])

    return df_clean
     
```

## Resumen acciones Cardinalidad
**Campos reducidos**: Funder, Instaler, Lga, Scheme_management.
**Campos eliminados**: Ward, wpt_name

```python

def Cardinalidad(df):

    df_agrupado = df.copy()

    # 1. DICCIONARIOS DE AGRUPACIÓN (Funciones internas)
    def agrupar_funder(nombre):
        if pd.isna(nombre): return 'Unknown'
        name = str(nombre).lower()
        if name in ['0', 'unknown', 'not known', 'none', '-']: return 'Unknown'
        if any(w in name for w in ['gov', 'ministry', 'council', 'lga', 'serikali', 'municipal', 'district', 'halmashauri', 'water board']): return 'Government'
        if any(w in name for w in ['church', 'catholic', 'roman', 'rc ', 'mission', 'diocese', 'kkkt', 'mosque', 'islam', 'kanisa', 'baptist', 'anglican', 'pentecost']): return 'Religious'
        if any(w in name for w in ['unicef', 'unice', 'world bank', 'w.b', 'w0rld', 'danida', 'vision', 'tasaf', 'oxfam', 'wateraid', 'hesawa', 'rwssp', 'amref', 'dwsp', 'tcrs', 'care', 'fund', 'red cross']): return 'International_NGO'
        if any(w in name for w in ['germany', 'japan', 'netherlands', 'nethalan', 'holland', 'china', 'usaid', 'finland', 'swedish', 'italy', 'british', 'belgian', 'kuwait', 'france']): return 'Foreign_Government'
        return 'Private_Local_Other'

    def agrupar_installer(nombre):
        if pd.isna(nombre): return 'Unknown'
        name = str(nombre).lower()
        if name in ['0', 'unknown', 'not known', 'none', '-']: return 'Unknown'
        if any(w in name for w in ['dwe', 'rwe', 'mwe', 'gov', 'ministry', 'council', 'lga', 'serikali', 'municipal', 'district', 'halmashauri', 'water board', 'idara', 'wizara', 'department']): return 'Government'
        if any(w in name for w in ['church', 'catholic', 'roman', 'rc ', 'mission', 'diocese', 'kkkt', 'mosque', 'islam', 'kanisa', 'baptist', 'anglican', 'pentecost', 'kkt']) or name == 'rc': return 'Religious'
        if any(w in name for w in ['unicef', 'unice', 'world bank', 'w.b', 'danida', 'vision', 'tasaf', 'oxfam', 'wateraid', 'hesawa', 'rwssp', 'amref', 'dwsp', 'tcrs', 'care', 'fund', 'red cross', 'twesa', 'acra']): return 'International_NGO'
        if any(w in name for w in ['germany', 'japan', 'netherlands', 'holland', 'china', 'usaid', 'finland', 'swedish', 'italy', 'british', 'belgian', 'kuwait', 'france', 'norad', 'jica', 'koica', 'wachina']): return 'Foreign_Government'
        return 'Private_Local_Other'

    def agrupar_scheme(scheme):
        if pd.isna(scheme): return 'Other'
        scheme = str(scheme)
        if scheme in ['VWC', 'WUG', 'WUA', 'SWC']: return 'Community'
        elif scheme in ['Water authority', 'Water Board', 'Parastatal']: return 'Government'
        elif scheme in ['Company', 'Private operator', 'Trust']: return 'Private'
        else: return 'Other'

    # 2. APLICAR LAS AGRUPACIONES AL DATAFRAME
    
    # TCampos:
    df_agrupado['funder_grouped'] = df_agrupado['funder'].apply(agrupar_funder)
    df_agrupado['installer_grouped'] = df_agrupado['installer'].apply(agrupar_installer)
    df_agrupado['scheme_grouped'] = df_agrupado['scheme_management'].apply(agrupar_scheme)

    # LGA (Rural vs Urbano)
    lga_str = df_agrupado['lga'].astype(str).str.lower()
    df_agrupado['lga_type'] = np.where(lga_str.str.contains('urban'), 'Urban',
                              np.where(lga_str.str.contains('rural'), 'Rural', 'Other'))

    columnas_a_borrar = ['funder', 'installer', 'scheme_management', 'lga', 'ward', 'wpt_name']
    df_agrupado = df_agrupado.drop(columns=columnas_a_borrar)
    

    return df_agrupado
```

# Conversión a numérica

Primero se va a empezar transformando los float64 a **int64** que se puedan hacer. Tabién transformar date_recorded dejando solo el año

```python
def transformacion_numeros(df):

    df_clean = df.copy()

    columnas_a_entero = ['population', 'construction_year']
    
    for col in columnas_a_entero:
        if col in df_clean.columns:
            df_clean[col] = df_clean[col].astype('int64')
            
    if 'date_recorded' in df_clean.columns:
        df_clean['date_recorded'] = pd.to_datetime(df_clean['date_recorded'])
        
        df_clean['year_recorded'] = df_clean['date_recorded'].dt.year.astype('int64')
        
        df_clean = df_clean.drop(columns=['date_recorded'])

    return df_clean
    
```

Ahora se aplica **One_Hot_Encoding** a: 

```python
def OneHot_Encoding(df):

    df_clean = df.copy()

    columnas_para_ohe = [
        'basin', 'region', 'public_meeting', 'permit',
        'extraction_type_class', 'management_group', 'payment', 
        'quality_group', 'quantity', 'source_type', 'waterpoint_type_group',
        'funder_grouped', 'installer_grouped', 'scheme_grouped', 'lga_type'
    ]

    cols_ohe_existentes = [col for col in columnas_para_ohe if col in df_clean.columns]
    df_clean = pd.get_dummies(df_clean, columns=cols_ohe_existentes, drop_first=True, dtype=int)

    return df_clean
     
```
